### A.2.1. Скаляры, векторы, матрицы и тензоры

````python
import torch
import numpy as np

# создаём 0D тензор (скаляр) из целого числа Python
tensor0d = torch.tensor(1)

# создаём 1D тензор (вектор) из списка Python
tensor1d = torch.tensor([1, 2, 3])

# создаём 2D тензор из вложенного списка Python
tensor2d = torch.tensor([[1, 2], 
                         [3, 4]])

# создаём 3D тензор из вложенного списка Python
tensor3d_1 = torch.tensor([[[1, 2], [3, 4]], 
                           [[5, 6], [7, 8]]])

# создаём 3D тензор из массива NumPy
ary3d = np.array([[[1, 2], [3, 4]], 
                  [[5, 6], [7, 8]]])
tensor3d_2 = torch.tensor(ary3d)  # Копирует массив NumPy
tensor3d_3 = torch.from_numpy(ary3d)  # Делит память с массивом NumPy
````

````python
ary3d[0, 0, 0] = 999
print(tensor3d_2) # остаётся неизменным
````

````python
print(tensor3d_3) # изменяется из-за общего доступа к памяти
````

Исходный тензор **2×3**:

```
[[1, 2, 3],
 [4, 5, 6]]
```

**tensor2d.T** (транспонирование, 3×2):
```
[[1, 4],
 [2, 5],
 [3, 6]]
```
С этим совпало.

---

**tensor2d.matmul(tensor2d.T)** — умножаем (2×3) на (3×2) → (2×2):
```
[1×1 + 2×2 + 3×3,  1×4 + 2×5 + 3×6]   = [14, 32]
[4×1 + 5×2 + 6×3,  4×4 + 5×5 + 6×6]   = [32, 77]
```

Результат:
```
[[14, 32],
 [32, 77]]
```

## A.3. Представление моделей в виде графов вычислений

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/7.webp" width="800px">

````python
import torch.nn.functional as F

y = torch.tensor([1.0])  # истинная метка
x1 = torch.tensor([1.1]) # входной признак
w1 = torch.tensor([2.2]) # весовой параметр
b = torch.tensor([0.0])  # нейрон смещения

z = x1 * w1 + b          # чистый вход
a = torch.sigmoid(z)     # активация и выход

loss = F.binary_cross_entropy(a, y)
print(loss)
````

## A.4. Автоматическое дифференцирование упростилось

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="800px">

````python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b 
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)
````

````python
loss.backward()

print(w1.grad)
print(b.grad)
````
