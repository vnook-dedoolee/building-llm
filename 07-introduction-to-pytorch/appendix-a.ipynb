## А.3. Представление моделей в виде графов вычислений

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/7.webp" width="800px">

````python
import torch.nn.functional as F

y = torch.tensor([1.0])  # истинная метка
x1 = torch.tensor([1.1]) # входной признак
w1 = torch.tensor([2.2]) # весовой параметр
b = torch.tensor([0.0])  # нейрон смещения

z = x1 * w1 + b          # чистый вход
a = torch.sigmoid(z)     # активация и выход

loss = F.binary_cross_entropy(a, y)
print(loss)
````

## А.4. Автоматическое дифференцирование упростилось

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="800px">

````python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b 
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)
````

````python
loss.backward()

print(w1.grad)
print(b.grad)
````

### А.9.2. Обучение с использованием одного графического процессора

````python
X_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])

y_train = torch.tensor([0, 0, 0, 1, 1])

X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])

y_test = torch.tensor([0, 1])
````

````python
from torch.utils.data import Dataset


class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)
````

````python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0, # num_workers > 0 в Jupyter на Windows часто вызывает ошибку
    drop_last=True
)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False,
    num_workers=0
)
````

 ---

Ошибка: `RuntimeError: DataLoader worker (pid(s) ...) exited unexpectedly`

**Почему на Windows в Jupyter:**
Windows использует `spawn` для создания новых процессов (вместо `fork`, как в Linux). Каждый воркер заново импортирует модули и выполняет код ячейки. В Jupyter это приводит к конфликтам и падению воркеров.

```python
# num_workers=0 — на Windows в Jupyter многопроцессность падает с RuntimeError.
# Ставим 0, чтобы загрузка данных шла в главном процессе.
train_loader = DataLoader(
    dataset=train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0,
    drop_last=True,
)
```

 ---

````python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(

            # 1-й скрытый слой
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2-й скрытый слой
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # выходной слой
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
````

````python
import torch.nn.functional as F


torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device) 

optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

num_epochs = 3

for epoch in range(num_epochs):

    model.train()
    for batch_idx, (features, labels) in enumerate(train_loader):

        features, labels = features.to(device), labels.to(device)
        logits = model(features)
        loss = F.cross_entropy(logits, labels) # Функция потерь

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        ### ЛОГГИРОВАНИЕ
        print(f"Эпоха: {epoch+1:03d}/{num_epochs:03d}"
              f" | Пакет {batch_idx+1:03d}/{len(train_loader):03d}"
              f" | Потери на обучении/валидации: {loss:.2f}")

    model.eval()
    # Опциональная оценка модели
````

````python
def compute_accuracy(model, dataloader, device):

    model = model.eval()
    correct = 0.0
    total_examples = 0

    for idx, (features, labels) in enumerate(dataloader):

        features, labels = features.to(device), labels.to(device) # New

        with torch.no_grad():
            logits = model(features)

        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)

    return (correct / total_examples).item()
````

````python
compute_accuracy(model, train_loader, device=device)
````

````python
compute_accuracy(model, test_loader, device=device)
````

### А.9.3. Обучение с использованием нескольких графических процессоров

````python
# Copyright (c) Sebastian Raschka под лицензией Apache License 2.0 (см. LICENSE.txt).
# Источник: "Build a Large Language Model From Scratch"
#   - https://www.manning.com/books/build-a-large-language-model-from-scratch
# Код: https://github.com/rasbt/LLMs-from-scratch

import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# импорты:
import os
import platform
import torch.multiprocessing as mp
from torch.utils.data.distributed import DistributedSampler
from torch.nn.parallel import DistributedDataParallel as DDP
from torch.distributed import init_process_group, destroy_process_group


# функция инициализации распределённой группы процессов (1 процесс / GPU)
# это обеспечивает взаимодействие между процессами
def ddp_setup(rank, world_size):
    """
    Аргументы:
        rank: уникальный идентификатор процесса
        world_size: общее количество процессов в группе
    """
    # ранг машины, запускающей процесс с rank:0
    # здесь мы предполагаем, что все GPU находятся на одной машине
    os.environ["MASTER_ADDR"] = "localhost"
    # любой свободный порт на машине
    os.environ["MASTER_PORT"] = "12345"

    # инициализируем группу процессов
    if platform.system() == "Windows":
        # Отключаем libuv, так как PyTorch для Windows собран без его поддержки
        os.environ["USE_LIBUV"] = "0"
        # Пользователям Windows, возможно, придётся использовать "gloo" вместо "nccl" в качестве бэкенда
        # gloo: Facebook Collective Communication Library
        init_process_group(backend="gloo", rank=rank, world_size=world_size)
    else:
        # nccl: NVIDIA Collective Communication Library
        init_process_group(backend="nccl", rank=rank, world_size=world_size)

    torch.cuda.set_device(rank)


class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]


class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
            # 1-й скрытый слой
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2-й скрытый слой
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # выходной слой
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits


def prepare_dataset():
    X_train = torch.tensor([
        [-1.2, 3.1],
        [-0.9, 2.9],
        [-0.5, 2.6],
        [2.3, -1.1],
        [2.7, -1.5]
    ])
    y_train = torch.tensor([0, 0, 0, 1, 1])

    X_test = torch.tensor([
        [-0.8, 2.8],
        [2.6, -1.6],
    ])
    y_test = torch.tensor([0, 1])

    # Раскомментируйте эти строки, чтобы увеличить размер набора данных для запуска этого скрипта на 8 GPU:
    # factor = 4
    # X_train = torch.cat([X_train + torch.randn_like(X_train) * 0.1 for _ in range(factor)])
    # y_train = y_train.repeat(factor)
    # X_test = torch.cat([X_test + torch.randn_like(X_test) * 0.1 for _ in range(factor)])
    # y_test = y_test.repeat(factor)

    train_ds = ToyDataset(X_train, y_train)
    test_ds = ToyDataset(X_test, y_test)

    train_loader = DataLoader(
        dataset=train_ds,
        batch_size=2,
        shuffle=False,  # False из-за DistributedSampler ниже
        pin_memory=True,
        drop_last=True,
        # распределяем пакеты по GPU без перекрытия образцов:
        sampler=DistributedSampler(train_ds)  # НОВОЕ
    )
    test_loader = DataLoader(
        dataset=test_ds,
        batch_size=2,
        shuffle=False,
    )
    return train_loader, test_loader


#  обёртка
def main(rank, world_size, num_epochs):

    ddp_setup(rank, world_size)  # инициализируем группы процессов

    train_loader, test_loader = prepare_dataset()
    model = NeuralNetwork(num_inputs=2, num_outputs=2)
    model.to(rank)
    optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

    model = DDP(model, device_ids=[rank])  # оборачиваем модель в DDP
    # теперь основная модель доступна как model.module

    for epoch in range(num_epochs):
        # Устанавливаем сэмплер, чтобы каждая эпоха имела разный порядок перемешивания
        train_loader.sampler.set_epoch(epoch)

        model.train()
        for features, labels in train_loader:

            features, labels = features.to(rank), labels.to(rank)  # используем rank
            logits = model(features)
            loss = F.cross_entropy(logits, labels)  # Функция потерь

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            # ЛОГГИРОВАНИЕ
            print(f"[GPU{rank}] Эпоха: {epoch+1:03d}/{num_epochs:03d}"
                  f" | Размер пакета {labels.shape[0]:03d}"
                  f" | Потери на обучении/валидации: {loss:.2f}")

    model.eval()

    try:
        train_acc = compute_accuracy(model, train_loader, device=rank)
        print(f"[GPU{rank}] Точность на обучении", train_acc)
        test_acc = compute_accuracy(model, test_loader, device=rank)
        print(f"[GPU{rank}] Точность на тесте", test_acc)

    ####################################################
    except ZeroDivisionError as e:
        raise ZeroDivisionError(
            f"{e}\n\nЭтот скрипт рассчитан на 2 GPU. Вы можете запустить его так:\n"
            "CUDA_VISIBLE_DEVICES=0,1 python DDP-script.py\n"
            f"Или, чтобы запустить его на {torch.cuda.device_count()} GPU, раскомментируйте код на строках 103–107."
        )
    ####################################################

    destroy_process_group()  # корректно завершаем распределённый режим


def compute_accuracy(model, dataloader, device):
    model = model.eval()
    correct = 0.0
    total_examples = 0

    for idx, (features, labels) in enumerate(dataloader):
        features, labels = features.to(device), labels.to(device)

        with torch.no_grad():
            logits = model(features)
        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)
    return (correct / total_examples).item()


if __name__ == "__main__":
    # Этот скрипт может не работать для GPU > 2 из-за маленького набора данных
    # Запустите `CUDA_VISIBLE_DEVICES=0,1 python DDP-script.py`, если у вас GPU > 2
    print("Версия PyTorch:", torch.__version__)
    print("CUDA доступен:", torch.cuda.is_available())
    print("Количество доступных GPU:", torch.cuda.device_count())
    torch.manual_seed(123)

    # порождаем новые процессы
    # обратите внимание, что spawn автоматически передаст rank
    num_epochs = 3
    world_size = torch.cuda.device_count()
    mp.spawn(main, args=(world_size, num_epochs), nprocs=world_size)
    # nprocs=world_size порождает один процесс на GPU
````
