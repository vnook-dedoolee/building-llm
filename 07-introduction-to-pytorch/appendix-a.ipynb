### A.2.1. Скаляры, векторы, матрицы и тензоры

````python
import torch
import numpy as np

# создаём 0D тензор (скаляр) из целого числа Python
tensor0d = torch.tensor(1)

# создаём 1D тензор (вектор) из списка Python
tensor1d = torch.tensor([1, 2, 3])

# создаём 2D тензор из вложенного списка Python
tensor2d = torch.tensor([[1, 2], 
                         [3, 4]])

# создаём 3D тензор из вложенного списка Python
tensor3d_1 = torch.tensor([[[1, 2], [3, 4]], 
                           [[5, 6], [7, 8]]])

# создаём 3D тензор из массива NumPy
ary3d = np.array([[[1, 2], [3, 4]], 
                  [[5, 6], [7, 8]]])
tensor3d_2 = torch.tensor(ary3d)  # Копирует массив NumPy
tensor3d_3 = torch.from_numpy(ary3d)  # Делит память с массивом NumPy
````

````python
ary3d[0, 0, 0] = 999
print(tensor3d_2) # остаётся неизменным
````

````python
print(tensor3d_3) # изменяется из-за общего доступа к памяти
````

### A.2.3. Распространённые операции с тензорами в PyTorch

````python
tensor2d = torch.tensor([[1, 2, 3], 
                         [4, 5, 6]])
tensor2d
````

````python
tensor2d.shape
````

````python
tensor2d.reshape(3, 2)
````

````python
tensor2d.view(3, 2)
````

````python
tensor2d.T
````

````python
tensor2d.matmul(tensor2d.T)
````

````python
tensor2d @ tensor2d.T
````

 ---

Исходный тензор **2×3**:

```
[[1, 2, 3],
 [4, 5, 6]]
```

**tensor2d.T** (транспонирование, 3×2):
```
[[1, 4],
 [2, 5],
 [3, 6]]
```
С этим совпало.

---

**tensor2d.matmul(tensor2d.T)** — умножаем (2×3) на (3×2) → (2×2):
```
[1×1 + 2×2 + 3×3,  1×4 + 2×5 + 3×6]   = [14, 32]
[4×1 + 5×2 + 6×3,  4×4 + 5×5 + 6×6]   = [32, 77]
```

Результат:
```
[[14, 32],
 [32, 77]]
```
